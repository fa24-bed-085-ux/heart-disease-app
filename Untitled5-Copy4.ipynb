```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import joblib

file_path = r'C:\Users\Dell\Downloads\heart+disease\processed.cleveland.data'
df = pd.read_csv(file_path, header=None)

X = df[[0, 1, 4, 7]]
y = df[df.columns[-1]].apply(lambda x: 1 if x > 0 else 0)

model = RandomForestClassifier(random_state=42)
model.fit(X, y)

joblib.dump(model, 'heart_model.pkl')
print("Done!")
```
